## 1. Antagonistic Tendon Actuation

Cables can only **pull** (transmit tensile force). A single tendon cannot
produce bidirectional torque at a revolute joint. The solution is an
**antagonistic pair**: two tendons routed on opposite sides of a pulley
(Salisbury & Craig, 1982).

For a pulley of radius $r$ with cable tensions $T_1$ and $T_2$:

$$\tau = r \cdot (T_1 - T_2) \qquad 
  \kappa \propto T_1 + T_2$$

- **Differential tension** ($T_1 \neq T_2$) produces a net **torque**.
- **Co-contraction** ($T_1 + T_2$ large) increases apparent **stiffness**
  without changing the equilibrium angle (Kobayashi et al., 1998).

For an $n$-DOF system, Lee & Tsai (1991) proved that **$n + 1$ cables**
are the minimum required for full force closure (bidirectional torque
on all joints with positive cable tensions).

### Torque–Stiffness Decoupling

With two cables ($m = 2$) for one joint ($n = 1$), there is one degree
of redundancy ($r = m - n = 1$). This null-space freedom allows
**independent control of torque and stiffness**.

Given tension bounds $0 \leq T_i \leq T_\mathrm{max}$, the
feasible region in $(\tau, \kappa)$ space forms a **diamond** (rhombus):

- At **zero torque**, the full stiffness range $[0, \kappa_\mathrm{max}]$ is accessible.
- At **maximum torque**, stiffness is fixed (no co-contraction headroom).
- The vertical axis represents the **null-space direction**: changing
  stiffness without affecting torque.

## 2. Four-Bar Linkage Topologies

A **four-bar linkage** consists of four rigid bars connected by revolute
joints in a single closed loop with one DOF (Uicker et al., 2011).

Two topologically distinct arrangements arise when opposite sides are equal:

| Property | Parallelogram | Antiparallelogram |
|---|---|---|
| Side links | Parallel (uncrossed) | **Crossed** |
| Coupler motion | Pure translation | Complex rotation |
| ICR | At infinity (parallel links) | **Migrates** along centrode ellipse |
| Effective lever arm | Constant | **Configuration-dependent** |
| Transmission ratio | 1:1 (linear) | Nonlinear, amplifying |

The **antiparallelogram** is used as the elbow joint in the thesis robot
(Klein, 2023). Its migrating ICR produces a variable moment arm, requiring
a configuration-dependent Jacobian for accurate tendon force mapping
(Dijksman, 1977; McCarthy & Soh, 2010).

## 3. Tensegrity Structure Classification

A **tensegrity** structure ("tensile integrity") consists of isolated
compression members (struts) held in stable equilibrium by a continuous
tension network (cables). Skelton & de Oliveira (2009) classify
tensegrity systems by the maximum number of struts meeting at a single
node:

- **Class-1**: No strut–strut contact. Each node connects at most one
  strut to a tension network. Example: the classic 3-strut triangular
  prism (Snelson's *X-Piece*, 1948).

- **Class-2**: Up to two struts share a joint node. This permits rigid
  articulation while retaining tensegrity's compliance and lightweight
  properties. Example: the antiparallelogram X-joint used in the thesis
  manipulator (Fasquelle et al., 2020).

### Key Properties for Robotics

| Property | Mechanism |
|---|---|
| Inherent compliance | Tension network absorbs impacts passively (Bicchi & Tonietti, 2004) |
| High strength-to-weight | Struts carry only axial loads — no bending moments |
| Controllable stiffness | Cable pretension modulates stiffness without VSA hardware |
| Remote actuation | Motors at base, cables transmit force to distal joints |

## 4. Cable-Driven Parallel Mechanism (CDPM)

A **cable-driven parallel mechanism** consists of a movable platform
connected to a fixed base by $m$ cables. The static equilibrium is
described by the **structure matrix** $\mathbf{A}^T$
(Verhoeven, 2004; Bruckmann et al., 2008):

$$\mathbf{w} = \mathbf{A}^T \mathbf{t}$$

where $\mathbf{w}$ is the platform wrench, $\mathbf{t}$ the cable tension
vector, and $\mathbf{A}^T \in \mathbb{R}^{n \times m}$ has columns
determined by cable directions and attachment geometry.

Since cables can only pull ($\mathbf{t} \geq 0$), **wrench closure**
requires $m \geq n + 1$ cables (Gosselin & Grenier, 2011). The
redundancy ($r = m - n$) enables tension optimisation via the null space
(Pott et al., 2009).

The thesis robot's **wrist** is a 2-DOF CDPM with 3 active cables
($n = 2$, $m = 3$, $r = 1$) arranged at 120° intervals
(Nemoto et al., 2022). Additional 3 passive cables provide the
kinematic constraint of a spherical joint.
